# Sales Performance & Customer behaviour Analysis

```python
import pandas as pd
import numpy as np
```

```python
#Sales Dataset -1
sales1=pd.DataFrame({
    "order_id":["O1001","O1002","O1003","O1004"],
    "order_date":["2025-05-28","2025-05-29","2025-05-30","2025-05-29"],
    "product":["lapToP","Mouse","Keyboard","Laptop"],
    "category":["Technology","tech","Hardware","Technology"],
    "region":["South","north","west","south"],
    "quantity":[1,2,1,1],
    "price":[60000,800,1500,66000],
    "discount":[0.10,np.nan,0.05,0.10]
})
sales1
```

```python
sales2 = pd.DataFrame({
    "order_id": ["O1005", "O1006", "O1003"],  
    "order_date": ["2025-06-07", "2025-08-06", "2025-07-02"],
    "product": ["Printer", "Laptop", "Keyboard"],
    "category": ["Hardware", "Technology", "hardware"],
    "region": ["East", "West", "North"],
    "quantity": [1, 1, 1],
    "price": [12000, 62000, 1550],
    "discount": [0.00, 0.15, 0.05]
})

sales2
```

```python
sales3 = pd.DataFrame({
    "order_id": ["O1007", "O1008", "O1009"],
    "order_date": ["2025/09/06", "2025/11/18", "2025/03/30"], 
    "product": ["Tablet", "Mouse", "Chair"],
    "category": ["Tech", "technology", "Furniture"],
    "region": ["South", "East", "North"],
    "quantity": [2, 3, 1],
    "price": [30000, 850, 5000],
    "discount": [np.nan, 0.07, "40%"]
})

sales3
```

```python
sales_raw=pd.concat([sales1,sales2,sales3])
sales_raw
```

```python
sales_raw=sales_raw.reset_index(drop=True)
sales_raw
```

```python
sales_raw["region"]=sales_raw["region"].str.strip().str.lower().str.title()
sales_raw
```

```python
sales_raw.groupby('region').count()
```

```python
category_map={
    "technology": "Technology",
    "tech": "Technology",
    "hardware": "Hardware",
    "furniture": "Furniture"
}
```

```python
sales_raw["category"]=(
sales_raw["category"]
    .str.strip()
    .str.lower()
    .map(category_map)
)
```

```python
sales_raw["category"].unique()
```

```python
sales_raw.groupby("category")["price"].sum()
```

```python
#cleaning product names
sales_raw["product"]=sales_raw["product"].str.strip().str.title()
```

```python
sales_raw["product"].unique()
```

```python
sales_raw["order_date"] = pd.to_datetime(sales_raw["order_date"], format="mixed")
sales_raw.info()
```

```python
sales_raw.dtypes
```

```python
sales_raw["category"].unique()
```

```python
sales_raw["discount"] = sales_raw["discount"].replace("40%", 0.40).astype(float)
sales_raw["discount"] = sales_raw["discount"].fillna(0)
```

```python
sales_raw["revenue"] = sales_raw["quantity"] * sales_raw["price"] * (1 - sales_raw["discount"])
sales_raw["revenue"]
```

```python
#overall sales performance
total_revenue=sales_raw["revenue"].sum()
total_revenue
```

```python
#region wise revenue
region_sales=sales_raw.groupby("region")["revenue"].sum().sort_values(ascending=False)
region_sales
```

```python
#top selling products
product_sales=sales_raw.groupby("product")["revenue"].sum().sort_values(ascending=False)
product_sales
```

```python
#monthly sales trend
monthly_sales=sales_raw.resample("ME", on="order_date")["revenue"].sum()
monthly_sales
```

```python
#discount impact
sales_raw.groupby("discount")["revenue"].sum()
```

```python
category_sales = sales_raw.groupby("category")["revenue"].sum().sort_values(ascending=False)
category_sales
```

```python
import matplotlib.pyplot as plt

# Region sales
region_sales.plot(kind="bar", title="Revenue by Region")
plt.show()

# Category sales
category_sales.plot(kind="bar", title="Revenue by Category")
plt.show()

# Monthly trend
monthly_sales.plot(title="Monthly Revenue Trend")
plt.show()

# Top products
product_sales.plot(kind="bar", title="Revenue by Product")
plt.show()
```
