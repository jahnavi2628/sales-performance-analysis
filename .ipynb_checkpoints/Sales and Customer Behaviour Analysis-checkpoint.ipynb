# Sales Performance & Customer behaviour Analysis

```python
import pandas as pd
import numpy as np
```

```python
#Sales Dataset -1
sales1=pd.DataFrame({
    "order_id":["O1001","O1002","O1003","O1004"],
    "order_date":["2025-05-28","2025-05-29","2025-05-30","2025-05-29"],
    "product":["lapToP","Mouse","Keyboard","Laptop"],
    "category":["Technology","tech","Hardware","Technology"],
    "region":["South","north","west","south"],
    "quantity":[1,2,1,1],
    "price":[60000,800,1500,66000],
    "discount":[0.10,np.nan,0.05,0.10]
})
sales1
```

```python
sales2 = pd.DataFrame({
    "order_id": ["O1005", "O1006", "O1003"],  
    "order_date": ["2025-06-07", "2025-08-06", "2025-07-02"],
    "product": ["Printer", "Laptop", "Keyboard"],
    "category": ["Hardware", "Technology", "hardware"],
    "region": ["East", "West", "North"],
    "quantity": [1, 1, 1],
    "price": [12000, 62000, 1550],
    "discount": [0.00, 0.15, 0.05]
})

sales2
```

```python
sales3 = pd.DataFrame({
    "order_id": ["O1007", "O1008", "O1009"],
    "order_date": ["2025/09/06", "2025/11/18", "2025/03/30"], 
    "product": ["Tablet", "Mouse", "Chair"],
    "category": ["Tech", "technology", "Furniture"],
    "region": ["South", "East", "North"],
    "quantity": [2, 3, 1],
    "price": [30000, 850, 5000],
    "discount": [np.nan, 0.07, "40%"]
})

sales3
```

```python
sales_raw=pd.concat([sales1,sales2,sales3])
sales_raw
```

```python
sales_raw=sales_raw.reset_index(drop=True)
sales_raw
```

```python
sales_raw["region"]=sales_raw["region"].str.strip().str.lower().str.title()
sales_raw
```

```python
sales_raw.groupby('region').count()
```

```python
category_map={
    "technology": "Technology",
    "tech": "Technology",
    "hardware": "Hardware",
    "furniture": "Furniture"
}
```

```python
sales_raw["category"]=(
sales_raw["category"]
    .str.strip()
    .str.lower()
    .map(category_map)
)
```

```python
sales_raw["category"].unique()
```

```python
sales_raw.groupby("category")["price"].sum()
```

```python
#cleaning product names
sales_raw["product"]=sales_raw["product"].str.strip().str.title()
```

```python
sales_raw["product"].unique()
```

```python
sales_raw["order_date"] = pd.to_datetime(sales_raw["order_date"], format="mixed")
sales_raw.info()
```

```python
sales_raw.dtypes
```

```python
sales_raw["category"].unique()
```

```python
sales_raw["discount"] = sales_raw["discount"].replace("40%", 0.40).astype(float)
sales_raw["discount"] = sales_raw["discount"].fillna(0)
```

```python
sales_raw["revenue"] = sales_raw["quantity"] * sales_raw["price"] * (1 - sales_raw["discount"])
sales_raw["revenue"]
```

```python
#overall sales performance
total_revenue=sales_raw["revenue"].sum()
total_revenue
```

```python
#region wise revenue
region_sales=sales_raw.groupby("region")["revenue"].sum().sort_values(ascending=False)
region_sales
```

```python
#top selling products
product_sales=sales_raw.groupby("product")["revenue"].sum().sort_values(ascending=False)
product_sales
```

```python
#monthly sales trend
monthly_sales=sales_raw.resample("ME", on="order_date")["revenue"].sum()
monthly_sales
```

```python
#discount impact
sales_raw.groupby("discount")["revenue"].sum()
```

```python
category_sales = sales_raw.groupby("category")["revenue"].sum().sort_values(ascending=False)
category_sales
```

```python
import matplotlib.pyplot as plt

category_revenue = sales_raw.groupby("category")["revenue"].sum()

category_revenue.plot(kind="bar")
plt.title("Total Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.show()

```

```python
region_revenue = sales_raw.groupby("region")["revenue"].sum()

region_revenue.plot(kind="bar")
plt.title("Total Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.show()
```

```python
sales_raw["month"] = sales_raw["order_date"].dt.to_period("M")

monthly_revenue = sales_raw.groupby("month")["revenue"].sum()

monthly_revenue.plot(kind="line", marker="o")
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

```python
product_revenue = sales_raw.groupby("product")["revenue"].sum().sort_values(ascending=False)

product_revenue.plot(kind="bar")
plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.show()
```

```python
sales_raw["month"] = sales_raw["order_date"].dt.month
sales_raw["day"] = sales_raw["order_date"].dt.day
sales_raw["weekday"] = sales_raw["order_date"].dt.day_name()
```

```python
def price_bucket(p):
    if p < 2000:
        return "Low"
    elif p < 20000:
        return "Medium"
    else:
        return "High"

sales_raw["price_segment"] = sales_raw["price"].apply(price_bucket)
sales_raw[["category","month","day","weekday","revenue","price_segment"]]
```

# Business Insights & Conclusion

1. Revenue Concentration

Technology category contributes the majority of revenue (≈ ₹2.3 Lakhs), while Hardware and Furniture have significantly lower contribution.

Laptops alone generate the highest revenue among all products, indicating strong demand for high-value electronics.

2. Regional Performance

South region shows the highest revenue, followed by West, East, and North.

North and East regions have lower sales volume and revenue, indicating potential market expansion opportunities.

3. Product-Level Trends

High-priced products (Laptop, Tablet) dominate total revenue despite lower order counts.

Low-priced products (Mouse, Keyboard) have higher frequency but contribute less to overall revenue.

4. Discount Impact

Orders with moderate discounts (10–15%) generate higher revenue compared to very high discounts (40%), suggesting that extreme discounting may not be optimal.

5. Time-Based Patterns

Sales peak in May, August, and September, showing seasonal demand.

Weekday analysis shows most purchases occur on mid-week and weekends, useful for promotional planning.
